```python
import os
import csv
import numpy as np
from matplotlib import pyplot as plt
```

```python
# We select two files from the same benchmark which we compare the results from.
selected_dir = "../benchmarking_results/csv_results/"
# The baseline:
file1 = "netlib_benchmark_1.0e-4_PDLP.csv"
# The experiment results which we compare against the baseline: 
file2 = "netlib_benchmark_1.0e-4_PDLP+Nesterov_kappa=0.3_threshold=0.8_similarity_scaling=true_similarity_option=COSINE_SIMILARITY.csv"
```

```python

# Finding the instances with the largest difference in KKT passes
instances_differences = []

csv_1 = open(f"{selected_dir}/{file1}", "r")
csv_2 = open(f"{selected_dir}/{file2}", "r")
file_reader_1 = csv.reader(csv_1)
file_reader_2 = csv.reader(csv_2)

first_row = True
for line_1, line_2 in zip(file_reader_1, file_reader_2): 
    if first_row:
        first_row = False
    else:
        instance_name_1 = line_1[0].strip()
        instance_name_2 = line_2[0].strip()

        if instance_name_1 != instance_name_2:
            print("ERROR: Comparing different instances with each other")
            break

        # Extracting the constraint matrix density: 
        constraint_matrix_density_1 = float(line_1[7].strip())
        constraint_matrix_density_2 = float(line_2[7].strip())

        if constraint_matrix_density_1 != constraint_matrix_density_2:
            print("ERROR: Comparing different constraint matrix densities with each other")
            break

        kkt_passes_1 = int(line_1[4].strip()) 
        kkt_passes_2 = int(line_2[4].strip()) 
        kkt_passes_difference = kkt_passes_1 - kkt_passes_2

        time_1 = float(line_1[3].strip())
        time_2 = float(line_2[3].strip())
        time_difference = time_1 - time_2

        instance_difference_tuple = (instance_name_1, constraint_matrix_density_1, kkt_passes_difference, time_difference)
        instances_differences.append(instance_difference_tuple)
# Closing the files:  
csv_1.close()
csv_2.close()

# instances_differences.sort(key=lambda tup: tup[2]) # Sorting on diff of KKT passes:
instances_differences.sort(key=lambda tup: tup[3], reverse=True) # Sorting on diff of seconds:
instances_differences.sort(key=lambda tup: tup[1], reverse=True) # Sorting matrix density:

num_instances_shown = 100
print(f"Instance   K density   KKT improvement   Time improvement")

for entry in instances_differences[:num_instances_shown]:
    print(f"{entry[0]:9} {entry[1]:10.4f} {entry[2]:17} {entry[3]:16.3f} s")
```
